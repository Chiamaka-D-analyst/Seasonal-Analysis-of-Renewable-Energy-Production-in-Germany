```python
import pandas as pd

df = pd.read_csv("Germany_monthly_energy_data_2023.csv")
df["Month"] = pd.to_datetime(df["Month"])
df.head()
```

```python
pivot_df = df.pivot(index="Month", columns="Energy_Source", values="Energy_MWh")
pivot_df.head()
```

```python
pivot_df
```

```python
import matplotlib.pyplot as plt

pivot_df.plot()
plt.ylabel("Energy Production (MWh)")
plt.title("Monthly Energy Production by Source - Germany (2023)")
plt.show()
```

```python
plt.savefig("monthly_energy_trends_germany.png")
```
